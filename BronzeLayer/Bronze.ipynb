```python
import pyspark
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder\
        .master('local[*]')\
        .appName('nti_project')\
        .getOrCreate()
```

```python
tables = ['Categories', 'Customers' , 'Orders' , 'OrdersDetails' ,'Employees' , 'Product' , 'Shippers' , 'Suppliers']

base_path = 'hdfs://hadoop-namenode:9000/Data/raw_data/'
output_path = 'hdfs://hadoop-namenode:9000/Data/parquet/'

for table in tables:
    path = f'{base_path}{table}.csv'
    
    df = (spark.read
              .option("header", "true") 
              .option("delimiter", ",")        
              .option("encoding", "ISO-8859-1")       
              .option("multiLine", "true")      
              .csv(path))
    
    count = df.count()

    if count > 0:
        df.write.mode('overwrite').parquet(f'{output_path}{table}')
        print(f"✅ {table}: {count} rows saved in Parquet")
        
print("🎉 All files have been successfully converted to Parquet!")
```
